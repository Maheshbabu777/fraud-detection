```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import json
from scipy.stats import randint,uniform
from sklearn.model_selection import RandomizedSearchCV
```

```python
print(os.listdir('/kaggle/input/'))
```

```python
dataset_path = '/kaggle/input/datasets/shouldichangemyname/credfrauddata'
X_train = np.load(os.path.join(dataset_path, 'X_train_scaled.npy'))
X_test = np.load(os.path.join(dataset_path, 'X_test_scaled.npy'))
y_train = np.load(os.path.join(dataset_path, 'y_train.npy'))
y_test = np.load(os.path.join(dataset_path, 'y_test.npy'))

with open(os.path.join(dataset_path, 'feature_names.json'), 'r') as f:
    features = json.load(f)
```

```python
from imblearn.over_sampling import SMOTE

smote=SMOTE(random_state=42)
X_train_s,y_train_s=smote.fit_resample(X_train,y_train)
```

```python
print(y_train_s.sum())
```

```python
from lightgbm import LGBMClassifier
```

```python
lgbm_param_grid={
    'n_estimators': randint(100, 600),
    'max_depth': randint(3, 12),
    'num_leaves': randint(15, 150),
    'learning_rate': uniform(0.01, 0.29),
    'subsample': uniform(0.6, 0.4),
    'colsample_bytree': uniform(0.6, 0.4),
    'min_child_samples': randint(5, 100),
    'reg_alpha': uniform(0, 1),
    'reg_lambda': uniform(0, 1),
}
```

```python
lgbm_search=RandomizedSearchCV(
    estimator=LGBMClassifier(),
    n_jobs=-1,
    random_state=42,
    cv=3,
    n_iter=10,
    param_distributions=lgbm_param_grid,
    verbose=1,
    scoring='roc_auc'
)
```

```python
import gc
gc.collect()
```

```python
lgbm_search.fit(X_train_s,y_train_s)
```

```python
lgbm_search.best_params_
```

```python
from sklearn.metrics import f1_score,roc_auc_score
```

```python
test_pred=lgbm_search.best_estimator_.predict(X_test)
test_proba=lgbm_search.best_estimator_.predict_proba(X_test)[:,1]
```

```python
test_auc=roc_auc_score(y_test,test_proba)
test_f1=f1_score(y_test,test_pred)
print(test_auc)
print(test_f1)
```

```python
best_model = lgbm_search.best_estimator_

best_model.booster_.save_model('lgbm_model.txt')

clean_params = {k: (v.item() if hasattr(v, 'item') else v) for k, v in lgbm_search.best_params_.items()}
results = {
    "best_params": clean_params,
    "cv_auc": float(lgbm_search.best_score_),
    "test_auc": float(test_auc),
    "test_f1": float(test_f1),
}
with open('lgbm_results.json', 'w') as f:
    json.dump(results, f)
```
